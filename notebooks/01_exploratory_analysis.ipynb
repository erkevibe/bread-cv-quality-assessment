# Exploratory analysis of the generated bread-CV artifacts

This notebook is intentionally read-only: run `make experiment` first. It visualises the audited manifest, extracted features, and generated metrics without changing the frozen split.

```python
import json
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
audit = json.loads((root / 'reports/data_audit.json').read_text())
manifest = pd.read_csv(root / 'results/data/manifest.csv')
features = pd.read_csv(root / 'results/data/features.csv')
metrics = pd.read_csv(root / 'results/metrics/metrics.csv')
audit['size_mapping'], manifest['split'].value_counts(dropna=False)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
for split, group in features.groupby('split'):
    axes[0].scatter(group['width_px'], group['target_width'], label=split, alpha=0.75)
    axes[1].scatter(group['height_px'], group['target_height'], label=split, alpha=0.75)
axes[0].set(xlabel='width_px', ylabel='target_width')
axes[1].set(xlabel='height_px', ylabel='target_height')
axes[0].legend()
fig.suptitle('Pixel measurements versus instrument targets')
fig.tight_layout()
```

```python
test_metrics = metrics.query("split == 'test' and metric in ['mae', 'rmse', 'r2']")
test_metrics.pivot(
    index='method', columns=['target', 'metric'], values='value'
).round(3)
```
